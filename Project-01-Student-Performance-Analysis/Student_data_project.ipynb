Histogram

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("student_data.csv")
plt.hist(data["age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.title("Age Distribution")
plt.show()
```

Barchart

```python
df.groupby("sex")["g3"].mean().plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.title("Average Final Grade by Gender")
plt.show()
```

Line chart

```python
plt.plot(df["g1"].head(20))
plt.xlabel("Student")
plt.ylabel("Math Score")
plt.title("Student Math Scores")
plt.show()
```

Scatter plot

```python
plt.scatter(df["g1"], df["g3"])
plt.xlabel("G1 Grade")
plt.ylabel("G3 Grade")
plt.title("G1 vs G3")
plt.show()
```

```python
Q1 = df["g3"].quantile(0.25)
Q3 = df["g3"].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df["g3"] < Q1 - 1.5*IQR) | (df["g3"] > Q3 + 1.5*IQR)]
print(outliers)
```
