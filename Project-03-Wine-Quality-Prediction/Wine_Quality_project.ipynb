Correlation matrix

```python
corr=df.corr()
print(corr)
```

Correlation heatmap


```python
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True)
plt.show()
```

```python
high_corr=corr[(corr>0.5) | (corr<-0.5)]
print(high_corr)
```

Box plot

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
high_quality=df[df["quality"] >7]
high_quality
```

```python
df.sort_values("quality",ascending=False)
```

Histogram

```python
df["quality"].hist()
plt.xlabel("Quality")
plt.ylabel("Frequency")
plt.show()
```

Scatter plot

```python
plt.scatter(df["alcohol"],df["quality"])
plt.xlabel=("Alcohol")
plt.ylabel=("Quality")
plt.show()
```

Bar chart

```python
plt.figure()
df.groupby("quality")["alcohol"].mean().plot(kind="bar")
plt.xlabel = "Quality"
plt.ylabel="Average Alcohol"
plt.title("Average Alcohol by Wine Quality")
plt.show()
```

```python
sns.countplot(x="quality",data=df)
plt.show()
```
