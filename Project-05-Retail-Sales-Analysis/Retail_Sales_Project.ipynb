```python
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Median Sales:", df["Sales"].median())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
```

```python
print("Total Profit:", df["Profit"].sum())
print("Average Profit:", df["Profit"].mean())
print("Median Profit:", df["Profit"].median())
print("Maximum Profit:", df["Profit"].max())
print("Minimum Profit:", df["Profit"].min())
```

```python
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)
```

```python
region_profit = df.groupby("Region")["Profit"].sum()
print(region_profit)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
print(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].sum()
print(category_profit)
```

Bar chart


```python
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

Line chart

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.plot(kind="line")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

Histogram

```python
plt.hist(df["Sales"], bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

Scatter plot

```python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

Correlation

```python
df[["Sales", "Quantity", "Discount", "Profit"]].corr()
```

```python
sns.heatmap(df[["Sales", "Quantity", "Discount", "Profit"]].corr(),annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
top_profit = df.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(10)
print(top_profit)
```
